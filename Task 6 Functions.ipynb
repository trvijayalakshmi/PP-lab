```python
def gcd(a, b):
    while b:
        a, b = b, a % b
    return a

# Test the function
result = gcd(68,4)
print("GCD is ", result)
print("GCD of 64 and 4 is:", gcd(64, 4))   # Output: 4
print("GCD of 25 and 15 is:", gcd(25, 15)) # Output: 5
```

```python
def fibonacci_sequence(n):
    a, b = 0, 1
    count = 0
    while count < n:
        print(a, end=" ")
        a, b = b, a + b
        count += 1

# Driver code
terms = int(input("Enter the number of terms: "))
print("Fibonacci sequence:")
fibonacci_sequence(terms)
```

```python
def factorial(n):
    fact = 1
    for i in range(1, n+1):
        fact *= i
    return fact

print("Factorial of 5 is:", factorial(5))
```

```python
# function which return reverse of a string
def isPalindrome(s):
    return s == s[::-1]

s = "malayalam"
print(isPalindrome(s))
```

```python
def isPalindrome(st):
    for i in range(0, int(len(st)//2)):
        if st[i] != st[len(st)-i-1]:
            return False
    return True
 
s = "malayalam"
print(isPalindrome(s))
if isPalindrome(s):
    print("The string is a palindrome")
else:
    print("The string is not a palindrome")
```

```python
def prime(x, y):
    for i in range(x, y):
        if i == 0 or i == 1:
            continue
        else:
            for j in range(2, int(i/2)+1):
                if i % j == 0:
                    break
            else:
                print(i,end=" ")

# Driver program
starting_range = 6
ending_range = 30
prime(starting_range, ending_range)
```

```python
def decimal_to_binary(n):
    if n == 0:
        return "0"
    b = ""
    while n > 0:
        rem = n % 2
        b = str(rem) + b
        n = n // 2
    return b

num = 25
print("Binary:", decimal_to_binary(num))
```

```python
def harmonic_sum(n):
    total = 0.0
    for i in range(1, n + 1):
        total += 1 / i
    return total

# Example usage
n = int(input("Enter a number: "))
print(f"Harmonic sum of {n} = {harmonic_sum(n)}")
```

```python
# one line using sum() and comprehension:
n = int(input("Enter a number: "))
harmonic = sum(1/i for i in range(1, n+1))
print(harmonic)
```

```python
def cal_interest(principal, rate=5, time=3):
    amount = principal * (1 + rate/100) ** time
    interest = amount - principal
    return interest

interest = cal_interest(30000)
print("The compound interest is:", interest)
interest = cal_interest(30000, 10)
print("The compound interest is:", interest)
interest = cal_interest(30000, 10, 4)
print("The compound interest is:", interest)
```

```python
def add_numbers(*nums):
    total = 0
    for n in nums:
        total += n
    return total

print(add_numbers(10, 20))
print(add_numbers(1, 2, 3, 4, 5))
```

```python
def employee(name, **details):
    print("Name:", name)
    for key, value in details.items():
        print(key, ":", value)

employee("Kiran", salary=40000, department="IT", city="Bangalore")
```
