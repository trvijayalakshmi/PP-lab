### Lambda Functions

```python
Lambda Syntax: lambda arguments: expression
```

```python
pairs = [(1, 5), (3, 1), (6, 8), (2, 4)]
sorted_pairs = sorted(pairs, key=lambda x: x[1])
print(sorted_pairs)   
```

```python
maximum = lambda a, b: a if a > b else b
print(maximum(12, 8))  
```

Map Syntax:  map(lambda arguments: expression, iterable)

```python
nums = [2, 4, 6, 8]
tripled = list(map(lambda x: x*3, nums))
print(tripled)   # Expected output: [6, 12, 18, 24]
```

```python
numbers = [1, 2, 3, 4, 5]
squares = list(map(lambda x: x**2, numbers))
print(squares) 
```

Filter Syntax: filter(lambda arguments: condition, iterable)

```python
words = ["Apple", "Banana", "Avocado", "Mango", "Apricot"]
a_words = list(filter(lambda w: w.startswith("A"), words))
print(a_words)
```

```python
nums = [10, 15, 20, 25, 30]
evens = list(filter(lambda x: x % 2 == 0, nums))
print(evens)
```

Reduce Syntax: reduce(lambda arguments: expression, iterable)

```python
from functools import reduce

n = 5
factorial = reduce(lambda x, y: x * y, range(1, n+1))
print("Factorial of", n, "=", factorial)
```

```python
from functools import reduce

numbers = [10, 45, 2, 99, 23]
maximum = reduce(lambda x, y: x if x > y else y, numbers)
print(maximum)
```

### Recursive Functions

```python
def gcd(a, b):
    if b == 0:
        return a
    return gcd(b, a % b)

print(gcd(48, 18))  
```

```python
def factorial(n):
    if n == 0 or n == 1:   # Base case
        return 1
    else:                  # Recursive case
        return n * factorial(n - 1)

print(factorial(5))  # Output: 120
```

```python
def fibonacci(n):
    if n <= 1:
        return n
    else:
        return fibonacci(n-1) + fibonacci(n-2)

n_terms = 10
print("Fibonacci sequence:")
for i in range(n_terms):
    print(fibonacci(i))
```

```python
def is_palindrome(s):
    if len(s)<=1: return True
    if s[0]!=s[-1]: return False
    return is_palindrome(s[1:-1])
print(is_palindrome("madam"))  
print(is_palindrome("racecar"))
print(is_palindrome("python")) 
```

```python
def sum_of_digits(n):
    if n < 10:
        return n
    else:
        return n % 10 + sum_of_digits(n // 10)

print(sum_of_digits(12345))
```

```python
def power(base, exponent):
    if exponent == 0:
        return 1
    else:
        return base * power(base, exponent - 1)

print(power(3, 5)) 
```

```python
def binary(n):
    if n > 1:
        binary(n//2)
    print(n % 2,end = '')
    
n = int(input("enter the decimal number"))
binary(n)
```
