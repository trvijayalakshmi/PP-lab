### a) Dictionary is a Mapping Type

```python
students = {}

# Input names and marks of 5 students
for i in range(5):
    name = input(f"Enter name of student {i+1}: ")
    marks = int(input(f"Enter marks of {name}: "))
    students[name] = marks

# Find the student with the highest marks
topper = max(students, key=students.get)

print("\nStudents and their marks:", students)
print("Topper:", topper, "with", students[topper], "marks")
```

```python
squares = {x: x*x for x in range(1, 10) if x%2==0}
print(squares) 
```

```python
data = {"a": 40, "a": 20, "c": 10, "d": 10}

unique_data = {}
for key, value in data.items():
    if value not in unique_data.values():
        unique_data[key] = value

print("After Removing Duplicates:", unique_data)
```

```python
# Without Comprehension
original = {
  "United States": "New York", 
  "Italy": "Naples" 
}
print("Original Dict", original)
inverted_dict = {}
for key, value in original.items():
    inverted_dict[value] = key
    
print("Inverted Dict", inverted_dict)
```

```python
# with dictionary comprehension
original = {
  "United States": "New York", 
  "Italy": "Naples" 
}
inverted = {value: key for key, value in original.items()}
print(inverted)
```

```python
marks = {"Alice": 85, "Bob": 92, "Charlie": 78}

topper = max(marks, key=marks.get)
print("Topper is:", topper)
```

```python
text = "programming"

freq = {}
for ch in text:
    freq[ch] = freq.get(ch, 0) + 1

print("Character Count:", freq)
```

```python
text = "programming"
# print(text.count('a'))
freq = {ch: text.count(ch) for ch in text}
print(freq)
```

### b) Sets

```python
sentence = "Python is easy and Python is powerful"
words = sentence.split()
unique_words = set(words)
print("Unique words:", unique_words)
print("Number of unique words:", len(unique_words))
```

```python
text = "programming"
print("Unique characters:", set(text))
unique_chars = "".join(set(text))
print("Unique characters:", unique_chars)
```

```python
# Create two sets
set1 = {1, 2, 3, 4, 5}
set2 = {4, 5, 6, 7, 8}

# Finding common elements
common = set1.intersection(set2)

# Finding elements present only in the first set
only_in_first = set1.difference(set2)

# Displaying results
print("Set 1:", set1)
print("Set 2:", set2)
print("Common Elements:", common)
print("Elements only in Set 1:", only_in_first)
```

```python
# Write a program to find common values between two arrays

import numpy as np 
ar1 = np.array([12, 14, 15, 16, 17])
ar2 = np.array([2, 4, 5, 12, 15]) 
# Common values between two arrays 
print(np.intersect1d(ar1, ar2))

ar3 = ar1 + ar2
print(ar3)
```

```python
numbers = [1, 2, 2, 3, 4, 4, 5, 6, 6]
unique_numbers = list(set(numbers))
print("Unique numbers:", unique_numbers)
```
